```python
# CS5760 Natural Language Processing
# Homework 2
# Bigram Language Model

# Training corpus
corpus = [
    ["<s>", "I", "love", "NLP", "</s>"],
    ["<s>", "I", "love", "deep", "learning", "</s>"],
    ["<s>", "deep", "learning", "is", "fun", "</s>"]
]

print("Training Corpus:")

for sentence in corpus:
    print(" ".join(sentence))
```

```python
from collections import Counter

# Initialize counters
unigram_counts = Counter()
bigram_counts = Counter()

# Count unigrams and bigrams
for sentence in corpus:
    unigram_counts.update(sentence)

    for i in range(len(sentence) - 1):
        bigram = (sentence[i], sentence[i + 1])
        bigram_counts[bigram] += 1

# Display unigram counts
print("Unigram Counts:")
for word, count in unigram_counts.items():
    print(word, ":", count)

print("\nBigram Counts:")
for bigram, count in bigram_counts.items():
    print(bigram, ":", count)
```

```python
# Function to calculate MLE bigram probability
def bigram_probability(previous_word, next_word):
    bigram_count = bigram_counts[(previous_word, next_word)]
    previous_word_count = unigram_counts[previous_word]

    if previous_word_count == 0:
        return 0

    return bigram_count / previous_word_count


# Display bigram probabilities
print("Bigram Probabilities:")

for bigram, count in bigram_counts.items():
    previous_word, next_word = bigram

    probability = bigram_probability(previous_word, next_word)

    print(
        f"P({next_word} | {previous_word}) = "
        f"{probability:.4f}"
    )
```

```python
# Function to calculate the probability of a sentence
def sentence_probability(sentence):
    probability = 1.0

    for i in range(len(sentence) - 1):
        previous_word = sentence[i]
        next_word = sentence[i + 1]

        probability *= bigram_probability(previous_word, next_word)

    return probability


# Test sentences
s1 = ["<s>", "I", "love", "NLP", "</s>"]
s2 = ["<s>", "I", "love", "deep", "learning", "</s>"]

# Calculate sentence probabilities
p1 = sentence_probability(s1)
p2 = sentence_probability(s2)

print("Sentence Probabilities:")
print("S1:", " ".join(s1))
print(f"P(S1) = {p1:.4f}")

print()

print("S2:", " ".join(s2))
print(f"P(S2) = {p2:.4f}")
```

```python
# Compare the two sentence probabilities

print("Model Preference:")

if p1 > p2:
    print("The model prefers S1: <s> I love NLP </s>")
    print("Reason: S1 has a higher probability than S2.")

elif p2 > p1:
    print("The model prefers S2: <s> I love deep learning </s>")
    print("Reason: S2 has a higher probability than S1.")

else:
    print("Both sentences have the same probability.")
```
